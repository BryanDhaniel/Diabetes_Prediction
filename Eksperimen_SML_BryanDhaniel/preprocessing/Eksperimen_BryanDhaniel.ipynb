# Eksperimen MSML: Prediksi Diabetes

## 1. Data Loading
Membaca dataset raw.

```python
import pandas as pd

df = pd.read_csv('../diabetes_prediction_raw/diabetes_prediction_dataset.csv')
df.head()
```

## 2. Exploratory Data Analysis (EDA)
Melakukan analisis pada dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns

df.info()
df.describe()
```

```python
sns.countplot(x='diabetes', data=df)
plt.title('Distribusi Target (Diabetes)')
plt.show()
```

## 3. Data Preprocessing
Melakukan pembersihan data, encoding, dan persiapan.

```python
from sklearn.preprocessing import LabelEncoder

# Handle missing values if any
df.dropna(inplace=True)

# Encoding categorical variables
le = LabelEncoder()
df['gender'] = le.fit_transform(df['gender'])
df['smoking_history'] = le.fit_transform(df['smoking_history'])

# Menyimpan data preprocessing
df.to_csv('../diabetes_prediction_preprocessing/diabetes_prediction_dataset_clean.csv', index=False)
print("Data bersih telah disimpan.")
```
